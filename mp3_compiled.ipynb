**Complex Function Mapping (Im[f(z)] vs Re[z])**

```python

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 500)
y = 2

z = x + 1j*y
f = z**2

plt.plot(x, f.imag, label='Imaginary part')
plt.xlabel('Real part of z')
plt.ylabel('Imaginary part of f(z)')
plt.title('Complex Function Mapping: f(z) = z²')
plt.grid()
plt.legend()
plt.show()
```

**Finite Difference Method**

```python

import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**2 + 3*x + 2

x = float(input("Enter x: "))
h = 0.001

derivative = (f(x+h) - f(x)) / h

print("Numerical Derivative =", derivative)

# Generate x values for plotting
X = np.linspace(x-5, x+5, 100)

# Calculate numerical derivative at each point
Y = (f(X+h) - f(X)) / h

plt.plot(X, Y, label='Numerical Derivative')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title('Finite Difference Method')
plt.legend()
plt.grid()
plt.show()
```

**Complex Function Mapping (f(z) vs z)**

```python

import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 2*np.pi, 500)

z = np.exp(1j*t)
w = z**2

fig, ax = plt.subplots(1, 2, figsize=(11, 5))

ax[0].plot(z.real, z.imag)
ax[0].set_title('Complex Plane: z')
ax[0].set_xlabel('Real(z)')
ax[0].set_ylabel('Imaginary(z)')

ax[1].plot(w.real, w.imag)
ax[1].set_title('Complex Plane: f(z) = z²')
ax[1].set_xlabel('Real(f(z))')
ax[1].set_ylabel('Imaginary(f(z))')

for a in ax:
    a.grid()
    a.axis('equal')

plt.tight_layout()
plt.show()
```

**Fourier Transform (Discrete Fourier Transform)**

```python

import numpy as np
import matplotlib.pyplot as plt

N = 1000
T = 1.0/N

t = np.linspace(0, 1, N, endpoint=False)

# Signal containing two frequencies
signal = (np.sin(2*np.pi*50*t)
          + 0.5*np.sin(2*np.pi*120*t))

fft = np.fft.fft(signal)
freq = np.fft.fftfreq(N, T)

plt.subplot(2, 1, 1)
plt.plot(t, signal)
plt.title('Original Signal')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.grid()

plt.subplot(2, 1, 2)
plt.plot(freq[:N//2], np.abs(fft[:N//2])*2/N)
plt.title('Discrete Fourier Transform')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Amplitude')
plt.grid()

plt.tight_layout()
plt.show()
```

**2D Vector Rotation / Transformation Matrix**

```python

import numpy as np
import matplotlib.pyplot as plt

x = float(input("Enter x coordinate: "))
y = float(input("Enter y coordinate: "))

angle = float(input("Enter rotation angle: "))

theta = np.radians(angle)

R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta), np.cos(theta)]
])

V = np.array([x, y])

result = np.dot(R, V)

print("Original Vector:", V)
print("Rotated Vector:", result)

# Plotting
plt.quiver(0, 0, V[0], V[1],
           angles='xy', scale_units='xy',
           scale=1, color='blue', label='Original Vector')

plt.quiver(0, 0, result[0], result[1],
           angles='xy', scale_units='xy',
           scale=1, color='red', label='Rotated Vector')

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('2D Vector Rotation')
plt.legend()
plt.grid()
plt.axis('equal')
plt.show()
```

**Gaussian-Laguerre Quadrature**

```python

import numpy as np

def func(x):
    return 1/(x**2 + 1)

# Gaussian-Laguerre Quadrature

n = int(input("Enter number of integration points: "))

x, w = np.polynomial.laguerre.laggauss(n)

fx = func(x)

ans = np.sum(w * fx * np.exp(x))

print("Gaussian-Laguerre Quadrature")
print("Integral =", ans)
```

**Gaussian-Hermite Quadrature**

```python

import numpy as np

def func(x):
    return 1/(x**2 + 1)

# Gaussian-Hermite Quadrature

n = int(input("Enter number of integration points: "))

x, w = np.polynomial.hermite.hermgauss(n)

fx = func(x)

ans = np.sum(w * fx)

print("Gaussian-Hermite Quadrature")
print("Integral =", ans)
```
